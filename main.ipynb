```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

# Specify the subreddit you want to scrape
subreddit_url = url

# Set up the Chrome WebDriver
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
driver.get(subreddit_url)

# Accept Reddit's cookies if prompted (optional)
try:
    cookies_button = driver.find_element(By.XPATH, "//button[contains(text(), 'Accept All')]")
    cookies_button.click()
except:
    pass

# Scroll a few times to load more posts
scroll_pause_time = 2
scroll_count = 10
for _ in range(scroll_count):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(scroll_pause_time)

def scrape_posts():
    # Extract titles and flairs
    posts = driver.find_elements(By.CLASS_NAME, "block.font-semibold.text-neutral-content-strong.m-0.visited\\:text-neutral-content-weak.text-16.xs\\:text-18.mb-2xs.xs\\:mb-xs.overflow-hidden")
    flairs = driver.find_elements(By.CLASS_NAME, "flair-content")

    # Loop through the posts and extract data
    for i in range(len(posts)):
        # Check if the current index exists in both posts and flairs
        if i < len(posts) and i < len(flairs):
            title = posts[i].text
            flair = flairs[i].text
        else:
            title = "N/A"
            flair = "N/A"

        # Scroll to the element before clicking
        driver.execute_script("arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});", posts[i])
        time.sleep(1)  # Wait for the element to scroll into view

        # Click on the post using JavaScript
        driver.execute_script("arguments[0].click();", posts[i])
        time.sleep(2)  # Wait for the post page to load

        # Scrape all <p> tags inside the description div
        try:
            post_id_element = driver.find_element(By.XPATH, "//div[contains(@id, '-post-rtjson-content')]")
            paragraphs = post_id_element.find_elements(By.TAG_NAME, "p")
            description = "\n".join([p.text for p in paragraphs])
        except:
            description = "N/A"

        print(f"Post {i+1}:")
        print(f"Title: {title}")
        print(f"Flair: {flair}")
        print(f"Description: {description}")
        print("-" * 40)

        # Go back to the main subreddit page
        driver.back()
        time.sleep(2)  # Wait for the subreddit page to reload

    # Refresh the page to fetch new posts
    driver.refresh()

# Keep scraping continuously
while True:
    scrape_posts()
    time.sleep(600)  # Refresh every 30 seconds (adjust as needed)
```
